The majority of the dataset has a frequency less than 10, hence we can not manually single out the fake/troll locations from the dataset. In this situation we are prompted to use generative AI to help us extract unreal locations into a json file "fake_location.json"

```python
import json

with open("fake_locations.json", encoding="utf-8") as f:
    data = json.load(f)

fake_locations = data["fake_locations"]  # just a plain list of strings (before location cleaning (raw format))
fake_locations_after = data["fake_location_after"] #list of strings that were discovered after location cleaning
print(len(fake_locations))
```

```python
initial_length = len(df_users_sub)
df_users_sub = df_users_sub[~df_users_sub["Location"].isin(fake_locations)]
print(initial_length-len(df_users_sub),"fake locations were removed.")
```

```python
import re

import unicodedata

def clean_location(location):
    # Decompose accented chars into base letter + combining accent mark
    normalized = unicodedata.normalize('NFKD', location)
    # Strip out the combining accent marks, keep the base letters
    no_accents = ''.join(c for c in normalized if not unicodedata.combining(c))
    # Now safely strip anything that isn't a plain letter or whitespace
    return re.sub(r'[^a-zA-Z\s]', ' ', no_accents)

df_users_sub["Location"] = df_users_sub["Location"].apply(clean_location).str.title()
df_users_sub = df_users_sub[df_users_sub["Location"].str.isspace()==False]
df_users_sub.loc[df_users_sub["Location"]=="Saudi Araba","Location"] = "Saudi Arabia"
cols=["name","state_name","country_name"]
for col in cols:
    df_cities_sub[col] = df_cities_sub[col].apply(clean_location).str.title()
```

```python

initial_length = len(df_users_sub)
df_users_sub = df_users_sub[~df_users_sub["Location"].isin(fake_locations_after)]
print(initial_length-len(df_users_sub),"fake cleaned locations were removed.")
```

```python
import re

troll_patterns = [
    # laughing / meme filler
    r"\bxd\b", r"\blol\b", r"\blmao\b", r"\brofl\b", r"\bhaha+\b", r"\blolol+\b",
    r"\bidk\b", r"\bsmh\b",

    # deflecting / refusing to answer
    r"none of your business|nun ya|nobody'?s business|not telling",
    r"who cares|who gives a|does it matter|no idea|i don ?t know",
    r"wouldn'?t you like to know|why do you (want|wanna) know|take a guess|guess",
    r"classified|confidential|top secret|need to know basis",
    r"stalker|creep|why do you care",r"human",

    # vague / non-answers
    r"^\s*(here|there|somewhere|nowhere|none|unknown|n\s*/\s*a|earth|everywhere)\s*$",
    r"in your (dreams|mind|head|heart)", r"behind you", r"right (here|behind|next to)",r"fluffy",
    r"my (house|home|room|bed|couch|basement)\b", r"in front of (my|the) (computer|screen|pc)",
    r"the internet|cyberspace|online|in my computer",
    r"under (a|the|my) (bed|rock)", r"in (a|my) (box|closet|fridge)", r"in my pants",

    # absurdist / meme places
    r"candy ?land|rainbow|unicorn|chocolate ?land|potato ?land|cloud ?9|neverland",
    r"area ?51|the moon|mars|outer space|the void|dreamland|wonderland",
    r"your mom'?s? (house|basement)?", r"cardboard box",r"night",

    # anime/fictional worlds (expand as you find more)
    r"hueco mundo|soul society|konoha|hidden leaf|karakura|naruto",
    r"grand line|raftel|alola|hyrule|amestris|ivalice|eorzea|gaia\b",
    r"death city|magnolia town|fiore\b|sinnoh|hoenn|midgar|radiant garden",
    r"twilight town|traverse town|destiny islands|gensokyo|aincrad",
    r"rune midgard|neo venezia|namimori|yggdrasil|zanarkand",

    # pop-culture fictional worlds
    r"narnia|mordor|the shire|hogwarts|gallifrey|krypton|pandora|panem",
    r"middle ?earth|skyrim|azeroth|discworld|ankh ?morpork|isla nublar",
    r"the matrix\b",

    # numeric / placeholder junk
    r"^\s*\d+\s*$", r"^\s*[.\-_?!]+\s*$", r"^\s*n/a\s*$", r"^\s*\?+\s*$",

    # any string containing the word "know"
    r"(?i).*\bknow\b.*",
]

def is_probable_troll(loc):
    if pd.isna(loc):
        return False
    loc_l = str(loc).lower()
    return any(re.search(p, loc_l) for p in troll_patterns)

initial_length = len(df_users_sub)
removed = df_users_sub[df_users_sub["Location"].apply(is_probable_troll)]
df_users_sub = df_users_sub[~df_users_sub["Location"].apply(is_probable_troll)]
print(initial_length - len(df_users_sub), "fake/troll locations were removed.")
```

```python
country_priority = [
    "United States","Philippines","Indonesia","United Kingdom","Brazil","Canada",
    "India","Mexico","Australia","France","Germany","Russia","Poland","Malaysia",
    "Italy","Spain","Vietnam","Thailand","Egypt","Turkey","Argentina","Netherlands",
    "Saudi Arabia","Nigeria","South Korea","Japan","Sweden","Peru","Colombia","Chile",
    "Pakistan","Bangladesh","Morocco","Algeria","South Africa","Ukraine","Romania",
    "Portugal","Belgium","Switzerland","Austria","Greece","Czech Republic","Hungary",
    "Norway","Denmark","Finland","Ireland","New Zealand","Israel","United Arab Emirates",
    "Singapore","Hong Kong S A R ","Taiwan","China","Kazakhstan","Serbia","Croatia","Bulgaria",
    "Slovakia","Slovenia","Lithuania","Latvia","Estonia","Iceland","Luxembourg","Malta",
    "Cyprus","Tunisia","Libya","Jordan","Lebanon","Iraq","Iran","Kuwait","Qatar",
    "Bahrain","Oman","Yemen","Syria","Afghanistan","Sri Lanka","Nepal","Myanmar",
    "Cambodia","Laos","Mongolia","Kyrgyzstan","Uzbekistan","Turkmenistan","Tajikistan",
    "Azerbaijan","Georgia","Armenia","Belarus","Moldova","Albania","North Macedonia",
    "Bosnia And Herzegovina","Montenegro","Kosovo","Ghana","Kenya","Ethiopia",
    "Tanzania","Uganda","Cameroon","Ivory Coast","Senegal","Zimbabwe","Zambia",
    "Angola","Mozambique","Botswana","Namibia","Rwanda","Ecuador","Venezuela",
    "Bolivia","Paraguay","Uruguay","Costa Rica","Panama","Guatemala","Honduras",
    "El Salvador","Nicaragua","Dominican Republic","Cuba","Jamaica",
    "Trinidad And Tobago","Puerto Rico","Fiji Islands","Papua New Guinea","Brunei","Bhutan",
    "Maldives","Timor Leste","North Korea","Andorra","Monaco","San Marino",
    "Liechtenstein","Greenland","Bermuda","The Bahamas","Barbados","Haiti",
]
country_priority_rank = {country: i for i, country in enumerate(country_priority)}
default_rank = len(country_priority)
```

```python
from collections import Counter
import numpy as np

# ============================================
# SETUP — run this ONCE, before applying to your 153k rows
# ============================================
unique_countries = set(df_cities_sub["country_name"])
unique_states = set(df_cities_sub["state_name"])
unique_cities = set(df_cities_sub["name"])

country_first_index = df_cities_sub.groupby("country_name").apply(lambda g: g.index[0]).to_dict()
state_first_index = df_cities_sub.groupby("state_name").apply(lambda g: g.index[0]).to_dict()
city_first_index = df_cities_sub.groupby("name").apply(lambda g: g.index[0]).to_dict()

# state_to_country = df_cities_sub.drop_duplicates("state_name").set_index("state_name")["country_name"].to_dict()
# city_to_country = df_cities_sub.drop_duplicates("name").set_index("name")["country_name"].to_dict()
state_to_countries = df_cities_sub.groupby("state_name")["country_name"].apply(set).to_dict()
city_to_countries = df_cities_sub.groupby("name")["country_name"].apply(set).to_dict()

country_abbreviations = {
    "USA": "United States",
    "U S A": "United States",
    "US": "United States",
    "U S": "United States",   # "U.S." -> punctuation stripped -> "U S"
    "America": "United States",
    "UK": "United Kingdom",
    "U K": "United Kingdom",  # "U.K." -> punctuation stripped -> "U K"
    "UAE": "United Arab Emirates",
    "U A E": "United Arab Emirates"
}

# ============================================
# FUNCTIONS
# ============================================
def row_index(val,location_type):
    if location_type=="country":
        index = country_first_index[val]
    elif location_type=="state":
        index = state_first_index[val]
    else:
        index = city_first_index[val]
    return index

def show_result(val,location_type):
    if location_type=="country":
        result = val
    elif location_type=="state":
        result = state_to_countries[val]   # now returns a SET of countries
    else:
        result = city_to_countries[val]    # now returns a SET of countries
    return result

def check_isolation(location, cand):
    location = ' ' + location + ' '
    pos = location.find(cand)
    isolated = (location[pos-1]==" ") and (location[pos+len(cand)]==" ")
    return isolated

def most_recurring_country(countries_list):
    counter = Counter(countries_list)
    most_common_country, count = counter.most_common(1)[0]
    return most_common_country, count


def determine_country(location):
    candidates=[]
    candidates_countries=[]
    candidates_types=[]

    location_upper = location.upper()

    # check abbreviations first — must be isolated, since they're short
    # and easily false-matched inside unrelated words (e.g. "UK" in "Ukraine")
    for abbrev, full_country in country_abbreviations.items():
        if location_upper.find(abbrev)!=-1:
            if check_isolation(location_upper, abbrev):
                # print("isolated abbreviation detected.")
                return (full_country, row_index(full_country,"country"))

    for country in unique_countries:
        if location.find(country)!=-1:
            #check if country mention is isolated
            if check_isolation(location,country):
                # print("isolated country detected.")
                return (country,row_index(country,"country"))
            
            candidates.append(country)
            candidates_types.append("country")
            candidates_countries.append(country)
        
    for state in unique_states:
        if location.find(state)!=-1:
            for country in show_result(state,"state"):
                candidates.append(state)
                candidates_types.append("state")
                candidates_countries.append(country)
        
    for city in unique_cities:
        if location.find(city)!=-1:
            for country in show_result(city,"city"):
                candidates.append(city)
                candidates_types.append("city")
                candidates_countries.append(country)

    # print(candidates)
    # print(candidates_types)
    # print(candidates_countries)
    if len(candidates)==0:
        return None
    
    rec_country, country_count = most_recurring_country(candidates_countries)

    #check by country count
    if country_count >=2:
        # print("recurring country detected.")
        return(rec_country,row_index(rec_country,"country"))
    
    #else return largest string (in terms of length)
    max_len = max(len(x) for x in candidates)
    tied_indices = [i for i,x in enumerate(candidates) if len(x)==max_len]

    if len(tied_indices) > 1:
        # multiple candidates tied on length -> break tie using country priority list
        tied_indices.sort(key=lambda i: country_priority_rank.get(candidates_countries[i], default_rank))

    best_cand_ind = tied_indices[0]
    # print("country detected by longest match.")
    best_country = candidates_countries[best_cand_ind]
    return (best_country, row_index(best_country,"country"))
```

```python
import time

def get_country(location):
    result = determine_country(location)
    return result[0] if result is not None else None

start = time.time()
df_users_sub["Country"] = df_users_sub["Location"].apply(get_country)
elapsed = time.time() - start
print(f"time spent: {elapsed / 60:.1f} minutes")
```
